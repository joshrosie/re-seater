```python
import json

# Load meta_Books.json
with open('./Yelp_JSON_2018/yelp_academic_dataset_business.json', 'r') as f:
    meta_data = {}
    for item in map(json.loads, f):
        categories = item.get('categories', [])
        if not categories:
            categories = []
        if isinstance(categories, str):
            categories = [cat.strip() for cat in categories.split(',')]
        meta_data[item['business_id']] = categories

# Prepare mapping from original to new book item ids, ignoring the first line
item_map = []
with open('./dataset/item_list.txt', 'r') as f:
    next(f)  # Skip the first line
    for line in f:
        orig_id, new_id = line.strip().split(" ")
        item_map.append((orig_id, new_id))

# Write new mapping to JSON: {new_id: categories}
book_categories_json = {new_id: meta_data.get(orig_id, []) for orig_id, new_id in item_map}
with open('./dataset/id2cat.json', 'w') as out_f:
    json.dump(book_categories_json, out_f)
```
