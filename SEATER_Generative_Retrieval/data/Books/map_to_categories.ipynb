```python
import json

# Load meta_Books.json
with open('./raw_data/meta_Books.json', 'r') as f:
    books_meta = {item['asin']: item.get('category', []) for item in map(json.loads, f)}

# Prepare mapping from original to new book item ids
book_item_map = []
with open('./data/book_data/book_item_map.txt', 'r') as f:
    for line in f:
        orig_id, new_id = line.strip().split(",")
        book_item_map.append((orig_id, new_id))

# Write new mapping to JSON: {new_id: categories}
book_categories_json = {new_id: books_meta.get(orig_id, []) for orig_id, new_id in book_item_map}
with open('new_book_item_categories.json', 'w') as out_f:
    json.dump(book_categories_json, out_f)
```
