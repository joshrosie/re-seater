```python
import re
def extract_data(file_path):
    with open(file_path, 'r') as f:
        next(f) 
        data = [
            [int(row.split('\t')[0]), [int(item) for item in re.findall(r'\d+', row.split('\t')[1])], int(row.split('\t')[2])]
            for row in f]
    return data
    
#data_books = extract_data('data/Books/dataset/training_cut20.tsv')
#print(data_books[:10])

```
