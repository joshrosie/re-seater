```python
import json

# Load meta_Books.json
meta_data = {}
with open('./MINDsmall_train/news.tsv', 'r', encoding='utf-8') as f:
    for line in f:
        parts = line.strip().split('\t')
        if len(parts) > 1:
            news_id = parts[0]
            category = parts[1]
            meta_data[news_id] = [category]

# Prepare mapping from original to new book item ids, ignoring the first line
item_map = []
with open('./dataset/news2idx.tsv', 'r') as f:
    for line in f:
        orig_id, new_id = line.strip().split("\t")
        item_map.append((orig_id, new_id))

# Write new mapping to JSON: {new_id: categories}
book_categories_json = {new_id: meta_data.get(orig_id, []) for orig_id, new_id in item_map}
with open('./dataset_impressions_concatenated/id2cat.json', 'w') as out_f:
    json.dump(book_categories_json, out_f)
```
